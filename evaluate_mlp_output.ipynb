```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ymaze_alt_pred = pd.read_csv('/Users/racheliritani/Desktop/AD Project/mm-vae/results_DIET/CGMVAE_rawdata/mlp/val_ymaze_alt_predictions.csv')
ymaze_alt_pred
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(ymaze_alt_pred['predicted_label'])
ax[0].set_title('Predicted Ymaze Spontaneous Alterations')
ax[0].set_xbound(0, 40)
ax[1].hist(ymaze_alt_pred['true_label'])
ax[1].set_title('Actual Ymaze Spontaneous Alterations')
ax[1].set_xbound(0, 40)
```
