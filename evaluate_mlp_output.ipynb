## Evaluating by diet

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
ymaze_cfc_pred_diet = pd.read_csv('/Users/racheliritani/Desktop/AD Project/mm-vae/results_DIET/VanillaVAE_5dim/mlp_test/predicted_cfc_epoch4000.csv')
ymaze_cfc_pred_diet 
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(ymaze_cfc_pred_diet['predicted_cfc'])
ax[0].set_title('Predicted CFC')
ax[0].set_xbound(0, 100)
ax[1].hist(ymaze_cfc_pred_diet['true_cfc'])
ax[1].set_title('Actual CFC')
ax[1].set_xbound(0, 100)
```

```python
fig, axs = plt.subplots(2, 3, figsize=(10, 8))
fig.suptitle('CFC predictions by diet label')
axs[0, 0].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 0.0, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 0].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 0.0, 'true_cfc'], label='true', alpha=0.5)
axs[0, 0].set_title('Diet label = 0')
axs[0, 1].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 1.0, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 1].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 1.0, 'true_cfc'], label='true', alpha=0.5)
axs[0, 1].set_title('Diet label = 1')
axs[0, 2].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 2.0, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 2].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 2.0, 'true_cfc'], label='true', alpha=0.5)
axs[0, 2].set_title('Diet label = 2')
axs[1, 1].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 3.0, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[1, 1].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 3.0, 'true_cfc'], label='true', alpha=0.5)
axs[1, 1].set_title('Diet label = 3')
axs[1, 2].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 4.0, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[1, 2].hist(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 4.0, 'true_cfc'], label='true', alpha=0.5)
axs[1, 2].set_title('Diet label = 4')
```

```python
diet_0_acc = mean_squared_error(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 0.0, 'true_label'], ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 0.0, 'predicted_label'])
diet_1_acc = mean_squared_error(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 1.0, 'true_label'], ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 1.0, 'predicted_label'])
diet_2_acc = mean_squared_error(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 2.0, 'true_label'], ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 2.0, 'predicted_label'])
diet_3_acc = mean_squared_error(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 3.0, 'true_label'], ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 3.0, 'predicted_label'])
diet_4_acc = mean_squared_error(ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 4.0, 'true_label'], ymaze_cfc_pred_diet.loc[ymaze_cfc_pred_diet['diet'] == 4.0, 'predicted_label'])
```

```python
print('diet 0 mse ' + str(diet_0_acc))
print('diet 1 mse ' + str(diet_1_acc))
print('diet 2 mse ' + str(diet_2_acc))
print('diet 3 mse ' + str(diet_3_acc))
print('diet 4 mse ' + str(diet_4_acc))
```
