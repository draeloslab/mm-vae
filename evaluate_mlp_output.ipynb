## Predictions with cfc conditioning

```python
import seaborn as sns
from sklearn.metrics import mean_absolute_error
```

```python
cfc_pred_cond = pd.read_csv('/Users/racheliritani/Desktop/AD Project/mm-vae/results_DIET/CGMVAE_cfc_cond_5bins_5dim/mlp/predicted_cfc_epoch10000.csv')
cfc_pred_cond
```

```python
all_data = pd.read_csv('/Users/racheliritani/Desktop/AD Project/VAE-work/c-gmvae/nature_filtered_nonan_cfc.csv')
all_data
```

```python
diet_data = all_data.loc[all_data['MouseID'].isin(cfc_pred_cond['mouse_ids']), ['MouseID', 'Diet_num']]
diet_data
```

```python
combined_pred_diet = pd.merge(cfc_pred_cond, diet_data, left_on='mouse_ids', right_on='MouseID')
combined_pred_diet
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(cfc_pred_cond['predicted_cfc'])
ax[0].set_title('Predicted CFC')
ax[0].set_xbound(0, 100)
ax[1].hist(cfc_pred_cond['true_cfc'])
ax[1].set_title('Actual CFC')
ax[1].set_xbound(0, 100)
```

```python
fig, axs = plt.subplots(2, 3, figsize=(10, 8))
fig.suptitle('Cfc predictions by diet label')
axs[0, 0].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 1, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 0].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 1, 'true_cfc'], label='true', alpha=0.5)
axs[0, 0].set_title('Diet label = 1')
axs[0, 1].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 2, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 1].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 2, 'true_cfc'], label='true', alpha=0.5)
axs[0, 1].set_title('Diet label = 2')
axs[0, 2].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 3, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[0, 2].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 3, 'true_cfc'], label='true', alpha=0.5)
axs[0, 2].set_title('Diet label = 3')
axs[1, 1].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 4, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[1, 1].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 4, 'true_cfc'], label='true', alpha=0.5)
axs[1, 1].set_title('Diet label = 4')
axs[1, 2].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 5, 'predicted_cfc'], label='predicted', alpha=0.5)
axs[1, 2].hist(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 5, 'true_cfc'], label='true', alpha=0.5)
axs[1, 2].set_title('Diet label = 5')
```

```python
color_map = {
    '1': 'red', 
    '2': 'green', 
    '3': 'purple', 
    '4': 'blue', 
    '5': 'orange'
}
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(x=combined_pred_diet['Diet_num'], y=combined_pred_diet['predicted_cfc'], data=combined_pred_diet, palette=color_map, ax=ax[0])
ax[0].set_title('Predicted CFC distribution by diet')
ax[0].set_ylim(0, 100)
sns.boxplot(x=combined_pred_diet['Diet_num'], y=combined_pred_diet['true_cfc'], data=combined_pred_diet, palette=color_map, ax=ax[1])
ax[1].set_title('Actual CFC distribution by diet')
```

```python
cfc_1_acc = mean_absolute_error(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 1, 'true_cfc'], combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 1, 'predicted_cfc'])
cfc_2_acc = mean_absolute_error(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 2, 'true_cfc'], combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 2, 'predicted_cfc'])
cfc_3_acc = mean_absolute_error(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 3, 'true_cfc'], combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 3, 'predicted_cfc'])
cfc_4_acc = mean_absolute_error(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 4, 'true_cfc'], combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 4, 'predicted_cfc'])
cfc_5_acc = mean_absolute_error(combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 5, 'true_cfc'], combined_pred_diet.loc[combined_pred_diet['Diet_num'] == 5, 'predicted_cfc'])
```

```python
print('Diet 40 (1) absolute error: ' + str(cfc_1_acc))
print('Diet 20 (2) absolute error: ' + str(cfc_2_acc))
print('Diet 2D (3) absolute error: ' + str(cfc_3_acc))
print('Diet 1D (4) absolute error: ' + str(cfc_4_acc))
print('Diet Ad (5) absolute error: ' + str(cfc_5_acc))
```
