```python
fig, ax = plt.subplots(figsize=(14, 5))

palette_cfc = sns.color_palette('flare', n_colors=umap1_with_labels_df['cfc_labels'].nunique())
#palette_cfc = categorical_palette
palette_diet = sns.color_palette('viridis', n_colors=umap1_with_labels_df['diet_labels'].nunique())

sns.scatterplot(data=umap1_with_labels_df, x='UMAP1', y='y_jitter_1', hue='cfc_labels',
                palette=palette_cfc, alpha=0.8, ax=ax, s=150, legend=False)
sns.scatterplot(data=umap1_with_labels_df, x='UMAP1', y='y_jitter_2', hue='diet_labels',
                palette=palette_diet, alpha=0.8, ax=ax, s=150, legend=False)

diet_categories = sorted(umap1_with_labels_df['diet_labels'].unique())
diet_color_map = dict(zip(diet_categories, palette_diet))
diet_handles = [Line2D([0], [0], marker='o', color='w', label=diet_label_map[label],
                       markerfacecolor=color, markersize=10)
                for label, color in diet_color_map.items()]

legend1 = ax.legend(handles=diet_handles, title='Diet Labels', bbox_to_anchor=(1.01, 1), loc='upper left')

cfc_categories = sorted(umap1_with_labels_df['cfc_labels'].unique())
cfc_color_map = dict(zip(cfc_categories, palette_cfc))
cfc_handles = [Line2D([0], [0], marker='o', color='w', label=label,
                      markerfacecolor=color, markersize=10)
               for label, color in cfc_color_map.items()]

legend2 = ax.legend(handles=cfc_handles, title='CFC Labels', bbox_to_anchor=(1.01, 0.6), loc='upper left')

ax.add_artist(legend1)

ax.set_yticks([0.1, 0.3])
ax.set_yticklabels(['CFC', 'Diet'], fontsize=12)
ax.set_ylim(-0.5, 1.5) 
ax.set_ylabel('')
ax.set_xlabel('UMAP Dimension 1', fontsize=12)
plt.ylim(0, 0.4)
plt.title('Projection onto UMAP1 colored by diet intervention and CFC', fontsize=15)
plt.tight_layout(rect=[0, 0, 0.88, 1])
plt.show()
```
