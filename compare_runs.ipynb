```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
osd_notrain = pd.read_csv('/home/rachel/Desktop/mm-vae results/multiregion_adbxd_cr_final/multiregion_testnotrainosd_seed15/osd_over_epochs.csv')
osd_train = pd.read_csv('/home/rachel/Desktop/mm-vae results/multiregion_adbxd_cr_final/multiregion_testtrainosd_seed15/osd_over_epochs.csv')
```

```python
plt.plot(osd_notrain, label='train_osd=False')
plt.plot(osd_train, label='train_osd=True')
plt.title('OSD over epochs')
plt.legend()
```

```python
mseloss_notrain = pd.read_csv('/home/rachel/Desktop/mm-vae results/multiregion_adbxd_cr_final/multiregion_testnotrainosd_seed15/mse_losses.csv')
mseloss_train = pd.read_csv('/home/rachel/Desktop/mm-vae results/multiregion_adbxd_cr_final/multiregion_testtrainosd_seed15/mse_losses.csv')
```
